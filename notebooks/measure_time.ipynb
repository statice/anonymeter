# Measure time taken to construct the singling out attacks

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from anonymeter.evaluators import SinglingOutEvaluator
from anonymeter.evaluators.singling_out_evaluator import _evaluate_queries, _generate_singling_out_queries
```

### Data

```python
dataset_dir = "./datasets/"

ori = pd.read_csv(os.path.join(dataset_dir, "adults_train.csv"))
syn = pd.read_csv(os.path.join(dataset_dir, "adults_syn_ctgan.csv"))
control = pd.read_csv(os.path.join(dataset_dir, "adults_control.csv"))
```

### Construct the evaluator

```python
max_attempts = 100_000
evaluator = SinglingOutEvaluator(ori=ori, 
                                 syn=syn, 
                                 control=control,
                                 n_attacks=max_attempts,
                                 n_cols=12,
                                 max_attempts=max_attempts)
```

### Run only the part of the evaluation that is concerned with constructing & running queries

```python
mode = "multivariate"
```

```python
%%timeit -n 1 -r 3 -o

queries = _generate_singling_out_queries(
    df=evaluator._syn,
    n_attacks=evaluator._n_attacks,
    n_cols=evaluator._n_cols,
    mode=mode,
    max_attempts=evaluator._max_attempts,
)
_queries = _evaluate_queries(df=evaluator._ori, queries=queries)
```
